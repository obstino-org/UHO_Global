### Prepare mel filters for C++ code (NeMo ASR)

```python
import torchaudio

sample_rate = 16_000
n_fft = 512

melscale_fbanks = torchaudio.functional.melscale_fbanks(
    n_fft // 2 + 1, 0, sample_rate // 2, 80, sample_rate, "slaney", "slaney"
).numpy()

melscale_fbanks.tofile("melfilters.bin")
```
