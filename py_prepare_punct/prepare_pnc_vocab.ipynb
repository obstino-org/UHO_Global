# Vocab file conversion

This notebook converts all vocab files of PnC

### Instructions how to prepare vocabulary files for inference in application UHO Global

Once you have downloaded the NeMo PnC models using ngc commands above do the following:

- unzip PnC .nemo file,
- put vocabulary files (e.g. f92889b136d2433693cb9127e1aea218_vocab.txt) from that file into folder "/vocab_files"
- run script below

### 1 First convert all vocab files for PnC

```python
import io
import os

base_path = "/mnt/e/pnc"

base_path_vocab = os.path.join(base_path, "vocab_files")
base_path_out = os.path.join(base_path, "final")

files = os.listdir(base_path_vocab)

for filename in files:
    fullpath = os.path.join(base_path_vocab, filename)
    fullpath2 = os.path.join(base_path_out, filename)
    f1 = io.open(fullpath, "r")
    f2 = io.open(fullpath2, "w")
    lines_out = []
    count = 0
    while True:
        line = f1.readline()
        if not line:
            break

        while len(line) and (line[-1] == '\r' or line[-1] == '\n'):
            line = line[0:-1]
        
        line_out = ""
        
        if line.startswith("[") and line.endswith("]"):
            line_out = line
        elif line.startswith("##"):
            line_out = line[2:]
        else:
            line_out = "▁" + line
        
        line_out += " " + str(count) + "\n"
        count += 1

        lines_out.append(line_out)
    f1.close()
    f2.writelines(lines_out)
    f2.close()
```
